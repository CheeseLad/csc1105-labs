Assume that your corpus consists of the following (the text file `corpus.txt`is in the same folder as this notebook):

```
I love the boy.
I love the dog.

They talk to the girl. 
They talk to the dog.

```

Let's start by cleaning the data a little: 

````python
from collections import defaultdict
import numpy as np
import string


def clean_sentence(sentence): 
    '''
    Separate out punctuation into their own "words'
    arg:
        sentence (str): Sentence as a string (should not include new line character).
    return:
        cleaned_sentence (str): Sentence with all punctuation separated by whitespace. 
    '''
    chars = []
    for i in range(len(sentence)): 
        if sentence[i] in string.punctuation: 
            # If there's whitespace before the punctuation
            if sentence[i-1]!=" ":
                char = " {}".format(sentence[i])
            # If we're not at the end of the sentence and there's whitespace after the punctuation. 
            if i!=len(sentence)-1 and sentence[i+1]!=" ":
                char = "{} ".format(char)
            chars.append(char)
        else:
            chars.append(sentence[i])
        cleaned_sentence = ''.join(chars)    
    return cleaned_sentence 

def read_text_into_array(filename):
    '''
    Read text file into array (removing new line character).
    arg:
        filename (str): Filename as string.
    return:
        corpus_stripped (list): List of stripped sentences (no new line character). 
    '''
    with open(filename, 'r') as f:
        corpus_raw = f.readlines()
        # Strip the new line character from each line.     
        corpus_stripped = [line.strip() for line in corpus_raw]
        return corpus_stripped
        
corpus = read_text_into_array("corpus.txt")
corpus_clean = []
for sentence in corpus:
    corpus_clean.append(clean_sentence(sentence))
corpus_clean

````

## Questions

### 1. Write a program that computes and prints out the probability of all the words in the vocabulary (use the `corpus_clean` list of sentences)


````python
def question_1(corpus_clean):
  word_counts = {}
  total_counts = 0
  for sentence in corpus_clean:
    for word in sentence.split():
      if word not in string.punctuation:
        if word not in word_counts:
          word_counts[word] = 1
          total_counts += 1
        else:
          word_counts[word] += 1
          total_counts += 1
  return word_counts, total_counts

word_counts, total_counts = question_1(corpus_clean)

print(word_counts)
print(total_counts)

for word in word_counts:
  word_counts[word] = word_counts[word]/total_counts
  print(f"Probability ({word}): {word_counts[word]}")
````

### 2. Write a program that computes and prints out the probability of each sentence in the corpus using a unigram model.



````python
word_counts, total_counts = question_1(corpus_clean)

import math

for sentence in corpus_clean:
  sentence_probability = []
  equation = ""
  fraction = ""
  for word in sentence.split():
    if word not in string.punctuation:
      equation += f"P({word}) * "
      fraction += f"({word_counts[word]} / {total_counts}) * "
      sentence_probability.append(word_counts[word] / total_counts)
      
  sentence_probability_value = 0.0
  for item in sentence_probability:
    sentence_probability_value += math.log(item)
    
  print(f"\nSentence: {sentence}")
  print(f"Equation: {equation.strip(' * ')}")
  print(f"Fraction: {fraction.strip(' * ')}")
  print(f"Values: {sentence_probability}")
  print(f"Probability: {math.exp(sentence_probability_value):.10f}")

    

    
````

**Bigrams**

````python
word_counts, total_counts = question_1(corpus_clean)

import math

def bigram(corpus_clean):
  bigram_counts = {}
  
  for sentence in corpus_clean:
    new_words = []
    for word in sentence.split():
      if word not in string.punctuation:
        new_words.append(word)
  
    for i in range(len(new_words) - 1):
      bigram = new_words[i] + " " + new_words[i + 1]
      if bigram not in bigram_counts:
        bigram_counts[bigram] = 1
      else:
        bigram_counts[bigram] += 1

  return bigram_counts

word_counts, total_counts = question_1(corpus_clean)
bigram_word_counts = bigram(corpus_clean)

for sentence in corpus_clean:
  sentence_probability = []
  equation = ""
  fraction = ""
  new_words = []
  for word in sentence.split():
    if word not in string.punctuation:
      new_words.append(word)

  for i in range(len(new_words)):
    if i == 0:
      equation += f"P({new_words[i]}) * "
    else:  
      equation += f"P({new_words[i]} | {new_words[i - 1]}) * "

  fraction = f"({word_counts[new_words[0]]} / {total_counts}) * "
  sentence_probability.append(word_counts[new_words[0]] / total_counts)
  
  for i in range(len(new_words) - 1):
    current_bigram = f"{new_words[i]} {new_words[i + 1]}"    
    fraction += f"({str(bigram_word_counts[current_bigram])} / {word_counts[new_words[i]]}) * "
    sentence_probability.append(bigram_word_counts[current_bigram] / word_counts[new_words[i]])
      
  sentence_probability_value = 0.0
  for item in sentence_probability:
    sentence_probability_value += math.log(item)
    
  print(f"\nSentence: {sentence}")
  print(f"Equation: {equation.strip(' * ')}")
  print(f"Fraction: {fraction.strip(' * ')}")
  print(f"Values: {sentence_probability}")
  print(f"Probability: {math.exp(sentence_probability_value):.10f}")
    

    
````

````python
word_counts, total_counts = question_1(corpus_clean)

import math

def trigram(corpus_clean):
  trigram_counts = {}
  
  for sentence in corpus_clean:
    new_words = []
    for word in sentence.split():
      if word not in string.punctuation:
        new_words.append(word)
  
    for i in range(len(new_words) - 2):
      trigram = new_words[i] + " " + new_words[i + 1] + " " + new_words[i + 2]
      if trigram not in trigram_counts:
        trigram_counts[trigram] = 1
      else:
        trigram_counts[trigram] += 1

  return trigram_counts

word_counts, total_counts = question_1(corpus_clean)
bigram_word_counts = bigram(corpus_clean)
trigram_word_counts = trigram(corpus_clean)

for sentence in corpus_clean:
  sentence_probability = []
  equation = ""
  fraction = ""
  new_words = []
  for word in sentence.split():
    if word not in string.punctuation:
      new_words.append(word)

  equation = f"P({new_words[0]}) * "
  equation += f"P({new_words[1]} | {new_words[0]}) * "

  for i in range(2, len(new_words)):
      equation += f"P({new_words[i]} | {new_words[i-2]} {new_words[i-1]}) * "

  fraction = f"({word_counts[new_words[0]]} / {total_counts}) * "
  sentence_probability.append(word_counts[new_words[0]] / total_counts)
  
  for i in range(len(new_words) - 2):
    current_trigram = f"{new_words[i]} {new_words[i + 1]} {new_words[i + 2]}"    
    fraction += f"({str(trigram_word_counts[current_trigram])} / {bigram_word_counts[new_words[i] + ' ' + new_words[i + 1]]}) * "
    sentence_probability.append(trigram_word_counts[current_trigram] / bigram_word_counts[new_words[i] + ' ' + new_words[i + 1]])
      
  sentence_probability_value = 0.0
  for item in sentence_probability:
    sentence_probability_value += math.log(item)
    
  print(f"\nSentence: {sentence}")
  print(f"Equation: {equation.strip(' * ')}")
  print(f"Fraction: {fraction.strip(' * ')}")
  print(f"Values: {sentence_probability}")
  print(f"Probability: {math.exp(sentence_probability_value):.10f}")
    

    
````
